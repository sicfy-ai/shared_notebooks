```python
from collections import defaultdict
from tqdm import tqdm
# Holds cell values and how long to hold them (rowspans)
rowspan_map = defaultdict(lambda: ["", 0])

for table_element in tqdm(table_elements, desc="Pr  ocessing tables", unit="table"):
    rows = table_element.find_all("tr")
    table_data = []
    max_cols = 0

    for row_index, row in enumerate(rows):
        cells = row.find_all(["td", "th"])
        row_data = []
        col_index = 0

        while col_index < 100:  # prevent infinite loops
            # Handle carried-over rowspan data
            if rowspan_map[col_index][1] > 0:
                row_data.append(rowspan_map[col_index][0])
                rowspan_map[col_index][1] -= 1
                col_index += 1
                continue

            try:
                cell = cells.pop(0)
            except IndexError:
                break

            cell_text = cell.get_text(strip=True)
            colspan = int(cell.get("colspan", 1))
            rowspan = int(cell.get("rowspan", 1))

            for i in range(colspan):
                if i == 0:
                    row_data.append(cell_text)
                    if rowspan > 1:
                        rowspan_map[col_index] = [cell_text, rowspan - 1]
                else:
                    row_data.append("")
                col_index += 1

        max_cols = max(max_cols, len(row_data))
        table_data.append(row_data)

    # Normalize
    for row in table_data:
        while len(row) < max_cols:
            row.append("")

    # Markdown generation
    is_header = all(cell.strip() != "" for cell in table_data[0])
    markdown = []

    if is_header:
        markdown.append("| " + " | ".join(table_data[0]) + " |")
        markdown.append("| " + " | ".join(["---"] * max_cols) + " |")
        data_rows = table_data[1:]
    else:
        # Insert dummy headers if no header detected
        markdown.append("| " + " | ".join(["Col {}".format(i+1) for i in range(max_cols)]) + " |")
        markdown.append("| " + " | ".join(["---"] * max_cols) + " |")
        data_rows = table_data

    for row in data_rows:
        markdown.append("| " + " | ".join(row) + " |")

    print("\n".join(markdown))
    print("-------------------------------------------------------")
```

```python
from bs4 import BeautifulSoup
from tqdm import tqdm
from collections import defaultdict
import html

def extract_markdown_from_cell(cell):
    def recurse(node):
        if node.name in ["strong", "b"]:
            return f"**{recurse_children(node)}**"
        elif node.name in ["em", "i"]:
            return f"*{recurse_children(node)}*"
        elif node.name == "code":
            return f"`{recurse_children(node)}`"
        elif node.name == "a":
            href = node.get("href", "#")
            return f"[{recurse_children(node)}]({href})"
        elif node.name is None:  # NavigableString
            return html.unescape(node.string or "")
        else:
            return recurse_children(node)

    def recurse_children(tag):
        return "".join(recurse(child) for child in tag.children)

    return recurse(cell).strip()


def html_tables_to_markdown(html_content):
    soup = BeautifulSoup(html_content, "html.parser")
    table_elements = soup.find_all("table")

    for table_element in tqdm(table_elements, desc="Processing tables", unit="table"):
        rows = table_element.find_all("tr")
        table_data = []
        rowspan_map = defaultdict(lambda: ["", 0])
        max_cols = 0

        # Optional: extract and print caption
        caption = table_element.find("caption")
        if caption:
            print(f"**{caption.get_text(strip=True)}**\n")

        for row_index, row in enumerate(rows):
            cells = row.find_all(["td", "th"])
            row_data = []
            col_index = 0

            while col_index < 100:
                # Carry-over from rowspan
                if rowspan_map[col_index][1] > 0:
                    row_data.append(rowspan_map[col_index][0])
                    rowspan_map[col_index][1] -= 1
                    col_index += 1
                    continue

                if not cells:
                    break

                cell = cells.pop(0)
                cell_text = extract_markdown_from_cell(cell)
                colspan = int(cell.get("colspan", 1))
                rowspan = int(cell.get("rowspan", 1))

                for i in range(colspan):
                    content = cell_text if i == 0 else ""
                    row_data.append(content)
                    if rowspan > 1 and i == 0:
                        rowspan_map[col_index] = [content, rowspan - 1]
                    col_index += 1

            max_cols = max(max_cols, len(row_data))
            table_data.append(row_data)

        # Normalize all rows
        for row in table_data:
            while len(row) < max_cols:
                row.append("")

        # Header detection
        first_row = rows[0]
        has_th = any(cell.name == "th" for cell in first_row.find_all(["td", "th"]))
        if has_th:
            header = table_data[0]
            data_rows = table_data[1:]
        else:
            header = [f"Col {i+1}" for i in range(max_cols)]
            data_rows = table_data

        # Markdown rendering
        markdown = []
        markdown.append("| " + " | ".join(header) + " |")
        markdown.append("| " + " | ".join(["---"] * max_cols) + " |")
        for row in data_rows:
            markdown.append("| " + " | ".join(row) + " |")

        print("\n".join(markdown))
        print("-" * 60)

html_tables_to_markdown(table_content)
```
